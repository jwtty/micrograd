This tutorial walks through the pytorch basics doc at https://docs.pytorch.org/tutorials/beginner/basics/tensorqs_tutorial.html.

```python
# Tensors are similar to NumPy’s ndarrays, 
# except that tensors can run on GPUs or other hardware accelerators. 
# Tensors and NumPy arrays can often share the same underlying memory, no need to copy data.
# Tensors are also optimized for automatic differentiation
import torch
import numpy as np
```

```python
# Initializing a tensor

# Directly from data
data = [[1,2],[3,4]]
x_data = torch.tensor(data)
print(type(data), type(x_data), x_data)
print("-------------")

# From a numpy array
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
print(np_array, x_np, type(np_array))
print("-------------")

# From another tensor
x_ones = torch.ones_like(x_data) # retains the properties of x_data
x_rand = torch.rand_like(x_data, dtype=torch.float) # overrides the datatype of x_data
print(x_ones, x_rand)
print("-------------")

# With random or constant values
shape = (2,3)
print(type(shape))
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)
print(rand_tensor, ones_tensor, zeros_tensor)
```

```python
# Attributes of a Tensor

tensor = torch.rand(2, 3)
print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
```

```python
# Operations on Tensors

# By default, tensors are created on the CPU. 
# We need to explicitly move tensors to the accelerator using .to method 
# (after checking for accelerator availability).
if torch.accelerator.is_available():
    tensor = tensor.to(torch.accelerator.current_accelerator())

# Indexing and slicing
tensor = torch.ones(4, 3)
print(f"First row: {tensor[0]}")
print(f"First col: {tensor[:, 0]}")
print(f"Last col: {tensor[:, -1]}")
tensor[:, 1] = 0
print(tensor)

# Joining tensors
t1 = torch.cat([tensor, tensor, tensor], dim=1)
print(t1.shape, t1)
t0 = torch.cat([tensor, tensor, tensor], dim=0)
print(t0.shape, t0)
```

```python
# Arithmetic operations

# Matrix multiplication
y1 = tensor @ tensor.T # ``tensor.T`` returns the transpose of a tensor
y2 = tensor.matmul(tensor.T)

y3 = torch.rand_like(y1)
torch.matmul(tensor, tensor.T, out=y3)
print(torch.equal(y1, y2) and torch.equal(y1, y3))
print(y1.shape, y1)

# Aggregation
agg = tensor.sum()
# item() converts one-element tensor to python numerical value
agg_item = agg.item() 
print(agg_item, type(agg_item))

# In-place updates, functions end with _
# In-place operations save some memory, 
# but can be problematic when computing derivatives because of an immediate loss of history. 
# Hence, their use is discouraged.
print(f"Before: {tensor}")
tensor.add_(5)
print(f"After add_(5): {tensor}")
tensor.copy_(tensor)
print(f"After copy_(): {tensor}")
```

```python
# Tensor to numpy
t = torch.ones(5)
print(f"t: {t}; type: {type(t)}")
n = t.numpy()
print(f"n: {n}; type: {type(n)}")

# A change in tensor reflects in numpy array
t[0] = 2
print(f"t: {t}")
print(f"n: {n}")

print("---------------")

# Numpy to tensor
n = np.ones(5)
t = torch.from_numpy(n)
print(f"n: {n}; type: {type(n)}")
print(f"t: {t}; type: {type(t)}")

# Changes in numpy array reflect in tensor, too
np.add(n, 1, out=n)
print(f"n: {n}")
print(f"t: {t}")
```
